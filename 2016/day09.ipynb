```python
import math
import re
```

```python
day09_inputs = "inputs/day09.txt"
with open(day09_inputs, 'r') as file:
    day09_data = file.readlines()
day09_data = [line.strip() for line in day09_data]
```

# part1

```python
marker_re = re.compile(r"\((\d+)x(\d+)\)")
def decompress_marker(input_string):
    pos = 0
    while True:
        match = marker_re.search(input_string, pos)
        if not match:
            return input_string
        
        search_len, repeat_times = map(int, match.groups()) 
        match_start, match_end = match.span()
        
        string_to_repeat = input_string[match_end:match_end+search_len]
        
        input_string = (
            input_string[:match_start]
            + string_to_repeat * repeat_times
            + input_string[match_end+search_len:]
        )
        pos = match_start + search_len*repeat_times

initial_input_string = day09_data[0]
output = decompress_marker(initial_input_string)
# print(output)
print(f"file has a decompressed length of {len(output)} vs. an original length of {len(initial_input_string)}")
```

# part2

```python
marker_re = re.compile(r"\((\d+)x(\d+)\)")
def decompress_string_length(input_string):
    total_length = 0
    curr_loc = 0
    
    while curr_loc < len(input_string):
        match = marker_re.match(input_string, curr_loc)
        if match:
            search_len, repeat_times = map(int, match.groups()) 
            match_start, match_end = match.span()
            
            string_to_repeat = input_string[match_end:match_end+search_len]
            
            # recursively calculate the length
            segment_len = decompress_string_length(string_to_repeat)
            total_length += segment_len * repeat_times
            
            curr_loc = match_end + search_len
        else:
            total_length += 1
            curr_loc += 1
            
    return total_length

input_string = day09_data[0]
output = decompress_string_length(input_string)
print(f"file has a decompressed length of {output} vs. an original length of {len(input_string)}")
```
