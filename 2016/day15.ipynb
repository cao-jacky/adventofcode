```python
import re

from collections import deque
```

```python
day15_inputs = "inputs/day15.txt"
with open(day15_inputs, 'r') as file:
    day15_data = file.readlines()
day15_data = [el.strip() for el in day15_data]
```

```python
discs_list = []
for line in day15_data:
    disc_num = int(re.search(r"#(\d+)", line).group(1))
    disc_total_posns = int(re.search(r"has (\d+) positions", line).group(1))
    disc_starting_posn = int(re.search(r"at position (\d+)", line).group(1))
    
    discs_list.append((disc_num, disc_total_posns, disc_starting_posn))
discs = tuple(discs_list)
```

```python
def disc_posns_at_time(time, disc):
    disc_num, total_posns, start_posn = disc
    pos = (start_posn + time) % total_posns
    return (disc_num, total_posns, pos)
```

```python
def discs_traversal(discs):
    starting_time = 0
    curr_capsule_loc = 0

    found = False
    while not found:
        starting_time += 1
        
        curr_time = starting_time
        capsule_path = []
        for disc in discs:
            curr_disc_posn = disc_posns_at_time(curr_time, disc)
            if curr_disc_posn[2] == 0:
                capsule_path.append((disc[0], curr_time))
            curr_time += 1
                
            if len(capsule_path) == len(discs):
                found = True
                return starting_time-1
```

```python
part1 = discs_traversal(discs)
print(f"part1: capsule has traversed the paths from starting time {part1}")
```

```python
new_discs_list = list(discs)
new_discs_list.append((7, 11, 0))
new_discs = tuple(new_discs_list)

part2 = discs_traversal(new_discs)
print(f"part2: capsule has traversed the paths from starting time {part2}")
```
