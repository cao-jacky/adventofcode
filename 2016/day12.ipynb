```python
day12_inputs = "inputs/day12.txt"
with open(day12_inputs, 'r') as file:
    day12_data = file.readlines()
day12_data = [el.strip() for el in day12_data]
```

```python
def instructions_loop(instructions, registers):
    i = 0
    while 0 <= i < len(instructions):
        instruction = instructions[i]
        primary_instruction = instruction[0]
        referenced_variable = instruction[-1]
        
        if "cpy" == primary_instruction:
            # the referenced value could be to another register or just a value
            referenced_value = registers[instruction[1]] if instruction[1] in registers else int(instruction[1])
            registers[referenced_variable] = referenced_value
            i += 1
        elif "inc" == primary_instruction:
            registers[referenced_variable] += 1
            i += 1
        elif "dec" == primary_instruction:
            registers[referenced_variable] -= 1
            i += 1
        elif "jnz" == primary_instruction:
            referenced_variable = instruction[1]
            # same here, another register, or a value
            referenced_value = registers[referenced_variable] if referenced_variable in registers else int(referenced_variable)
            if referenced_value != 0:
                offset = instruction[-1]
                i += int(offset)
            else:
                i += 1
    return registers
```

```python
instructions = [line.split() for line in day12_data]

registers = {"a": 0, "b": 0, "c": 0, "d": 0}
part1 = instructions_loop(instructions, registers)
print("part1", part1)
```

```python
registers = {"a": 0, "b": 0, "c": 1, "d": 0}
part2 = instructions_loop(instructions, registers)
print("part2", part2)
```
