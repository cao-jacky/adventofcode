```python
import numpy as np
```

```python
designers_fav_num = 10
```

```python
empty_floor_plan = np.zeros([6,9])
rows, cols = np.indices(empty_floor_plan.shape)

intermediate_floor_plan = cols*cols + 3*cols + 2*cols*rows + rows + rows*rows
intermediate_floor_plan += designers_fav_num

binary_floor_plan = np.vectorize(np.binary_repr)(intermediate_floor_plan)
count_ones_floor_plan = np.strings.count(binary_floor_plan, '1')

floor_plan = np.where(count_ones_floor_plan % 2 == 0, "1", "0")
floor_plan = floor_plan.astype(int)

print(binary_floor_plan)
print(count_ones_floor_plan)
print(floor_plan)
```
