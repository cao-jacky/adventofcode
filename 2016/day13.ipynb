```python
import numpy as np

from collections import deque
```

```python
designers_fav_num = 1362
```

```python
def eqn(x,y):
    return x*x + 3*x + 2*x*y + y + y*y
```

```python
def generate_next_states(state):
    x, y = state
    next_states = []
    
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    for dx, dy in directions:
        new_x, new_y = (x + dx, y + dy)
        
        if not (new_x >= 0 and new_y >= 0):
            continue 
        
        # check if new position is valid
        eqn_val = eqn(new_x, new_y)
        intermediate_val = eqn_val + designers_fav_num
        binary_val = np.binary_repr(intermediate_val)
        count_ones_val = np.strings.count(binary_val, '1')
        free_space = (True if count_ones_val % 2 == 0 else False)
        if free_space:
            next_states.append((new_x, new_y))
    return next_states
            
```

```python
def bfs(initial_state, end_state):
    # track positions and steps taken
    queue = deque([(initial_state, 0)])
    
    visited = set()
    visited.add(initial_state)
    
    while queue:
        state, steps = queue.popleft()
        
        if state == end_state:
            return steps
        
        for next_state in generate_next_states(state):
            if next_state not in visited: 
                visited.add(next_state)
                queue.append((next_state, steps + 1))
    return -1
```

# part1

```python
starting_point = (1,1)
destination_point = (31,39)

result = bfs(starting_point, destination_point)
print(f"part1: fewest number of steps required to move from {starting_point} to {destination_point} is {result}")
```

# part2

```python
def bfs(initial_state, max_steps):
    # track positions and steps taken
    queue = deque([(initial_state, 0)])
    
    visited = set()
    visited.add(initial_state)
    
    while queue:
        state, steps = queue.popleft()
        
        if steps == max_steps:
            return len(visited)
        
        for next_state in generate_next_states(state):
            if next_state not in visited: 
                visited.add(next_state)
                queue.append((next_state, steps + 1))
    return -1
```

```python
starting_point = (1,1)
max_steps = 50

result = bfs(starting_point, max_steps)
print(f"part2: there are {result} distinct locations that can be reached in at most, {max_steps} steps")
```
