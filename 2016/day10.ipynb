```python
import math
import re
```

```python
day10_inputs = "inputs/day10.txt"
with open(day10_inputs, 'r') as file:
    day10_data = file.readlines()
day10_data = [line.strip() for line in day10_data]
```

```python
def sort_by_values_len(dict):
    dict_len= {key: len(value) for key, value in dict.items()}
    import operator
    sorted_key_list = sorted(dict_len.items(), key=operator.itemgetter(1), reverse=True)
    sorted_dict = [{item[0]: dict[item [0]]} for item in sorted_key_list]
    return sorted_dict
```

```python
initial_values_instructions = [instruction for instruction in day10_data if "value" in instruction]
rest_of_instructions = [instruction for instruction in day10_data if "value" not in instruction]
```

```python
bots = {}

responsible_search = {17, 61} # part1

for ivi in initial_values_instructions:
    initial_task = ivi.split(" ")
    bot_num = int(re.findall(r"bot (\d+)", ivi)[0])
    microchip_value = initial_task[1]
    bots.setdefault(f"bot{bot_num}", []).append(int(microchip_value))    
bots = {key: sorted(value) for key, value in bots.items()}
print("initial values assigned to the bots are:")
print(bots)
print("")

remaining_instructions = rest_of_instructions.copy()
while remaining_instructions:
    processed = []
    
    for instruction in remaining_instructions:
        bot_num = int(re.findall(r"bot (\d+)", instruction)[0])
        
        if "gives" in instruction and len(bots.get(f"bot{bot_num}", [])) == 2:
            low_reciepient = re.search(r"low to\s+(?:output|bot)\s+(\d+)", instruction).group(1)
            high_receipient = re.search(r"high to\s+(?:output|bot)\s+(\d+)", instruction).group(1)
            
            low_reciepient_type = re.search(r"low to (\w+) (\d+)", instruction).group(1)
            high_reciepient_type = re.search(r"high to (\w+) (\d+)", instruction).group(1)
            
            curr_bot_chips = bots[f"bot{bot_num}"]
            if responsible_search.issubset(curr_bot_chips):
                print(f"part1: bot {bot_num} contains chips {responsible_search}")
            
            bots.setdefault(f"{low_reciepient_type}{low_reciepient}", []).append(curr_bot_chips[0])
            bots.setdefault(f"{high_reciepient_type}{high_receipient}", []).append(curr_bot_chips[1])
            
            bots[f"bot{bot_num}"].clear()
            bots = {key: sorted(value) for key, value in bots.items()}
            
            processed.append(instruction)
    
    for instr in processed:
        remaining_instructions.remove(instr)
    
    if not processed:
        break
    
print("")
bots = dict(sorted(bots.items()))
print("final values assigned to the bots and outputs are:")
print(bots)

print("")
outputs = [bots[f"output{i}"][0] for i in range(3)]
part2_calc = math.prod(outputs)
print(f"part2: multiplying outputs 0, 1, and 2 gives a value of {part2_calc}")
```
