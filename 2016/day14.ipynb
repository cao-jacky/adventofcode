```python
import hashlib
import re
```

# part1

```python
prearranged_salt = "yjdafjpo"

to_search_five_row = []
found_keys = []
integer_index = 0

while len(found_keys) < 64:
    salt_index = f"{prearranged_salt}{integer_index}"
    salt_index_hashed = hashlib.md5(salt_index.encode()).hexdigest()
    
    for search_limit, quintuplet_to_search, original_index in to_search_five_row:
        if integer_index > search_limit:
            continue
        if quintuplet_to_search in salt_index_hashed:
            found_keys.append(original_index)
    
    search_three_row = re.search(r'(.)\1\1', salt_index_hashed)
    if search_three_row:
        matched_triplet = search_three_row.group()
        quintuplet = matched_triplet[0]*5
        to_search_five_row.append([integer_index+1000, quintuplet, integer_index])
    
    integer_index += 1

print(f"part1: index {found_keys[63]} produces my 64th one-time pad key")
```
